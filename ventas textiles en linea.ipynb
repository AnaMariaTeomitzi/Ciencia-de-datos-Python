```python
import pandas as pd

# Cargar el archivo de Excel a un DataFrame
# Reemplaza 'ventas_tienda.xlsx' por el nombre EXACTO de tu archivo entre comillas
df_mis_ventas = pd.read_excel('/ventasjulio2026.xlsx')

# Revisar las primeras 5 filas para verificar que cargó bien
print("--- PRIMERAS FILAS DE MI EXCEL ---")
print(df_mis_ventas.head())

# cuenta cuentas colunmas hay en nuestro excel
filas, columnas = df_mis_ventas.shape

print(f"Tu tabla tiene un total de {filas} ventas registradas y {columnas} columnas.")

# 1. Ver qué columnas tiene la tabla
print("--- COLUMNAS DE LA TABLA ---")
print(df_mis_ventas.columns.tolist())

# 2. Ver cuántas ventas hay en total y qué tipo de datos son
print("\n--- RESUMEN GENERAL ---")
print(df_mis_ventas.info())

# Ver si hay celdas vacías en alguna columna
print("--- DATOS FALTANTES POR COLUMNA ---")
print(df_mis_ventas.isnull().sum())

# Ver las primeras 3 filas de la tabla
print("\n--- MUESTRA DE DATOS ---")
print(df_mis_ventas.head(3))

# Muestra las primeras 5 filas de una sola columna
print(df_mis_ventas['Título de la publicación'].head())

# 1. MÉTRICAS GENERALES (KPIs)
# ==========================================
# Sustituye 'Ingresos por ventas (MXN)' y 'Unidades' por los nombres exactos de tus columnas
total_ingresos = df_mis_ventas['Ingresos por productos (MXN)'].sum()
total_unidades = df_mis_ventas['Unidades'].sum()
ticket_promedio = df_mis_ventas['Ingresos por productos (MXN)'].mean()

print("--- RESUMEN EJECUTIVO ---")
print(f"Total vendido: ${total_ingresos:,.2f} MXN")
print(f"Total unidades vendidas: {total_unidades:,} piezas")
print(f"Venta promedio por transacción: ${ticket_promedio:,.2f} MXN")

# ==========================================
# 2. TOP 5 PRODUCTOS MÁS VENDIDOS
# ==========================================
print("\n--- TOP 5 PRODUCTOS (POR UNIDADES) ---")
top_unidades = (
    df_mis_ventas.groupby('Título de la publicación')['Unidades']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)
print(top_unidades)

print("\n--- TOP 5 PRODUCTOS (POR INGRESOS) ---")
top_ingresos = (
    df_mis_ventas.groupby('Título de la publicación')['Ingresos por productos (MXN)']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)
print(top_ingresos)

import matplotlib.pyplot as plt

# 1. Obtener los 10 productos más vendidos (en unidades)
top_10_unidades = (
    df_mis_ventas.groupby('Título de la publicación')['Unidades']
    .sum()
    .sort_values(ascending=True)  # Ascendente para que el producto top quede hasta arriba
    .tail(10)                     # Seleccionamos los últimos 10
)

# 2. Crear la figura y ajustar tamaño
plt.figure(figsize=(10, 6))

# 3. Dibujar las barras horizontales (kind='barh')
ax = top_10_unidades.plot(kind='barh', color='#3498db', edgecolor='black')

# 4. Personalizar títulos y ejes
plt.title('Top 10 Productos Más Vendidos (Unidades)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Unidades Vendidas', fontsize=12)
plt.ylabel('')  # Dejamos el eje Y sin etiqueta para no recargar la imagen

# 5. Colocar el valor numérico exacto al final de cada barra
ax.bar_label(ax.containers[0], fmt='%d', padding=5, fontsize=10, fontweight='bold')

# Ajustar márgenes para que los nombres largos no se corten
plt.tight_layout()

# Mostrar la gráfica
plt.show()
```
