```python
import pandas as pd
import numpy as np

# 1. Crear un dataset ficticio de ventas
datos = {
    'Producto': ['Sábanas', 'Toallas', 'Edredones', 'Sábanas', 'Toallas', 'Edredones'],
    'Ventas_Unidades': [15, 30, np.nan, 22, 18, 10],  # np.nan simula un dato faltante
    'Precio_Unitario': [250, 120, 450, 250, 120, 450]
}

df = pd.DataFrame(datos)

# 2. Exploración inicial
print("--- PRIMERAS FILAS ---")
print(df.head())

print("\n--- INFORMACIÓN Y DATOS NULOS ---")
print(df.isnull().sum())

# 3. Limpieza: Rellenar el valor faltante con la mediana de ventas
mediana_ventas = df['Ventas_Unidades'].median()
df['Ventas_Unidades'] = df['Ventas_Unidades'].fillna(mediana_ventas)

# 4. Transformación: Calcular el Ingreso Total
df['Ingreso_Total'] = df['Ventas_Unidades'] * df['Precio_Unitario']

# 5. Agrupación: Ventas e ingresos totales por producto
resumen = df.groupby('Producto')[['Ventas_Unidades', 'Ingreso_Total']].sum()

print("\n--- RESUMEN POR PRODUCTO ---")
print(resumen)
```

Limpieza de Datos: Manejo de Nulos
isnull().sum(): Cuenta cuántos valores nulos hay por columna.

fillna(valor): Rellena los huecos vacíos con el valor indicado.

mean(): Calcula el promedio aritmético.

median(): Encuentra el valor central (más robusto ante datos extremos).
