```python
import pandas as pd
import numpy as np

# Datos proporcionados por la empresa
datos_tienda = {
    'Producto': ['Laptop', 'Mouse', 'Teclado', 'Laptop', 'Mouse', 'Teclado'],
    'Categoria': ['Cómputo', 'Accesorios', 'Accesorios', 'Cómputo', 'Accesorios', 'Accesorios'],
    'Unidades_Vendidas': [5, 12, np.nan, 8, 15, 6],  # np.nan es un dato faltante
    'Precio_Unitario': [12000, 250, 450, 12000, 250, 450]
}

# Crear el DataFrame
df_ventas = pd.DataFrame(datos_tienda)
# 1. Exploración inicial Primera Solicitud
# Exploración: Muestra las primeras filas de la tabla y verifica cuántos valores nulos (vacíos) hay en total.
print("--- PRIMERAS FILAS ---")
print(df_ventas.head())
print("\n--- INFORMACIÓN Y DATOS NULOS ---")
print(df_ventas.isnull().sum())

# 2. Segunda Solicitud
# Limpieza: Rellena el valor faltante de Unidades_Vendidas usando la mediana de esa misma columna.
mediana_ventas = df_ventas['Unidades_Vendidas'].median()
df_ventas['Unidades_Vendidas'] = df_ventas['Unidades_Vendidas'].fillna(mediana_ventas)

# 3 Nueva Métrica: Crea una columna llamada Ingreso_Total que sea el resultado de multiplicar Unidades_Vendidas por Precio_Unitario.
df_ventas['Ingreso_Total'] = df_ventas['Unidades_Vendidas'] * df_ventas['Precio_Unitario']

# 4 Resumen: Calcula la suma global de Ingreso_Total agrupado por Categoria.
resumen = df_ventas.groupby('Categoria')[['Unidades_Vendidas', 'Ingreso_Total']].sum()
print("\n--- RESUMEN POR PRODUCTO ---")
print(resumen)


import matplotlib.pyplot as plt

# 1. Definir los datos para la gráfica desde nuestro resumen
categorias = resumen.index
ingresos = resumen['Ingreso_Total']

# 2. Crear la gráfica de barras
plt.figure(figsize=(7, 4))
plt.bar(categorias, ingresos, color=['#2b5c8f', '#d95f02'])

# 3. Personalizar la gráfica para que sea clara
plt.title('Ingreso Total por Categoría de Producto', fontsize=12, fontweight='bold')
plt.xlabel('Categoría', fontsize=10)
plt.ylabel('Ingresos ($)', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.5)

# 4. Mostrar la gráfica
plt.show()

```

groupby() Funciona igual que una tabla dinámica en Excel: junta o agrupa todas las filas que comparten la misma categoría (por ejemplo, 'Accesorios') y luego aplica una función matemática como la suma (sum) a las columnas numéricas."

"Un DataFrame es la tabla completa con filas y columnas (como una hoja de Excel). Una Serie es una sola columna seleccionada de esa tabla (como df_ventas['Unidades_Vendidas'])."

### 💡 Conceptos Clave de Manipulación de Datos
- **isnull().sum()**: Cuenta datos faltantes en cada columna.
- **fillna(valor)**: Rellena celdas vacías para evitar errores matemáticos.
- **median() vs mean()**: La mediana es el centro de los datos; el promedio es la suma dividida entre la cantidad.
- **groupby('columna')**: Agrupa datos por categorías para hacer resúmenes ejecutivos.
